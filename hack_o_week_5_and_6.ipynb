```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("student_data.csv")

print("Dataset Shape:", df.shape)
display(df.head())
```

```python
# Numerical columns
df.select_dtypes(include='number').columns
```

```python
marks = df[['G1', 'G2', 'G3']].iloc[0].values

print("Student Marks Vector:")
print(marks)
```

```python
matrix = df[['G1', 'G2', 'G3']].head(5).values

print("Marks Matrix:")
print(matrix)
```

```python
student1 = df[['G1', 'G2', 'G3']].iloc[0].values
student2 = df[['G1', 'G2', 'G3']].iloc[1].values

dot = np.dot(student1, student2)

print("Dot Product:", dot)
```

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

result = np.dot(A, B)

print(result)
```

```python
marks_data = df[['G1', 'G2', 'G3']]

covariance = np.cov(marks_data.T)

print("Covariance Matrix:")
print(covariance)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(covariance)

print("Eigenvalues:")
print(eigenvalues)
```

```python
x = 5

derivative = 2 * x

print("Derivative:", derivative)
```

```python
x = 5
y = 20
w = 2

prediction = w * x

gradient = 2 * (prediction - y) * x

print("Prediction:", prediction)
print("Gradient:", gradient)
```

```python
x = 5

u = 3*x + 2

du_dx = 3
dy_du = 2*u

dy_dx = dy_du * du_dx

print("Chain Rule Result:", dy_dx)
```

```python
x = 5
y = 20
w = 2

prediction = w * x

error = prediction - y

gradient = 2 * error * x

learning_rate = 0.01

new_w = w - learning_rate * gradient

print("Old Weight:", w)
print("Gradient:", gradient)
print("New Weight:", new_w)
```

```python
x = df['G2'].values
y = df['G3'].values

w = 1
b = 0

prediction = w * x + b

print("Actual G3:")
print(y[:10])

print("\nPredicted G3:")
print(prediction[:10])
```

```python
loss = np.mean((prediction - y)**2)

print("Loss:", loss)
```
