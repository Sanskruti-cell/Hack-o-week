```python
class StudentData:

    def __init__(self, dataframe):
        self.dataframe = dataframe

    def show_data(self):
        print(self.dataframe.head())

    def show_shape(self):
        print("Rows and Columns:", self.dataframe.shape)

    def show_columns(self):
        print("Columns:", self.dataframe.columns.tolist())


# Create object
student_data = StudentData(df)

student_data.show_data()
student_data.show_shape()
student_data.show_columns()
```

```python
import numpy as np

# Convert numerical data into NumPy array
arr = df.select_dtypes(include="number").to_numpy()

print("NumPy Array:")
print(arr)

# Vectorized operation
print("\nArray + 5:")
print(arr + 5)

# Broadcasting
print("\nArray * 2:")
print(arr * 2)

# Mean
print("\nColumn-wise Mean:")
print(np.mean(arr, axis=0))
```

```python
# Remove duplicates
df = df.drop_duplicates()

# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Fill missing numerical values
df[df.select_dtypes(include="number").columns] = \
    df.select_dtypes(include="number").fillna(
        df.select_dtypes(include="number").mean()
    )

# GroupBy
category = df.select_dtypes(include="object").columns[0]
numeric = df.select_dtypes(include="number").columns[0]

print("\nGroupBy Result:")
print(df.groupby(category)[numeric].mean())

# Sorting
print("\nTop 5:")
print(df.nlargest(5, numeric))
```

```python
# Create a small DataFrame for merging
extra_data = pd.DataFrame({
    category: df[category].unique(),
    "Category_ID": range(1, len(df[category].unique()) + 1)
})

merged_df = pd.merge(df, extra_data, on=category)

print(merged_df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Bar chart
df.groupby(category)[numeric].mean().plot(kind="bar")
plt.title("Average by Category")
plt.xlabel(category)
plt.ylabel(numeric)
plt.show()

# Histogram
plt.hist(df[numeric], bins=10)
plt.title("Distribution of Data")
plt.xlabel(numeric)
plt.ylabel("Frequency")
plt.show()

# Correlation Heatmap
sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create dashboard
fig = plt.figure(figsize=(16, 10))

fig.suptitle(
    "Student Performance Dashboard",
    fontsize=24,
    fontweight="bold"
)

# ================= KPI VALUES =================

total_students = len(df)
avg_g3 = df["G3"].mean()
highest_g3 = df["G3"].max()
avg_absences = df["absences"].mean()

# ================= KPI CARDS =================

ax1 = plt.subplot2grid((3, 4), (0, 0))
ax1.text(0.5, 0.5, f"Total Students\n{total_students}",
         ha="center", va="center", fontsize=18, fontweight="bold")
ax1.axis("off")

ax2 = plt.subplot2grid((3, 4), (0, 1))
ax2.text(0.5, 0.5, f"Average Final Grade\n{avg_g3:.2f}",
         ha="center", va="center", fontsize=18, fontweight="bold")
ax2.axis("off")

ax3 = plt.subplot2grid((3, 4), (0, 2))
ax3.text(0.5, 0.5, f"Highest Final Grade\n{highest_g3}",
         ha="center", va="center", fontsize=18, fontweight="bold")
ax3.axis("off")

ax4 = plt.subplot2grid((3, 4), (0, 3))
ax4.text(0.5, 0.5, f"Average Absences\n{avg_absences:.1f}",
         ha="center", va="center", fontsize=18, fontweight="bold")
ax4.axis("off")


# ================= G1 G2 G3 PERFORMANCE =================

ax5 = plt.subplot2grid((3, 4), (1, 0), colspan=2)

averages = [
    df["G1"].mean(),
    df["G2"].mean(),
    df["G3"].mean()
]

ax5.bar(["G1", "G2", "G3"], averages)

ax5.set_title("Average Grade Progress")
ax5.set_ylabel("Average Grade")


# ================= ABSENCES DISTRIBUTION =================

ax6 = plt.subplot2grid((3, 4), (1, 2), colspan=2)

ax6.hist(df["absences"], bins=10)

ax6.set_title("Absence Distribution")
ax6.set_xlabel("Absences")
ax6.set_ylabel("Number of Students")


# ================= G1 vs G3 =================

ax7 = plt.subplot2grid((3, 4), (2, 0), colspan=2)

ax7.scatter(df["G1"], df["G3"])

ax7.set_title("G1 vs Final Grade (G3)")
ax7.set_xlabel("G1 Grade")
ax7.set_ylabel("G3 Grade")


# ================= CORRELATION =================

ax8 = plt.subplot2grid((3, 4), (2, 2), colspan=2)

corr_columns = [
    "studytime",
    "failures",
    "absences",
    "G1",
    "G2",
    "G3"
]

corr = df[corr_columns].corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    ax=ax8
)

ax8.set_title("Performance Correlation")


plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
